# What does a 30-year mortgage cost?

The calculation from `demos/01_mortgage.rn`, one step per cell. Choose the **Rune (rnx)** kernel and run all cells; nothing is downloaded.

The monthly payment of a fixed-rate loan comes from the annuity formula.

```rune
fn payment(principal, yearly_rate, years) {
    let r = yearly_rate / 12.0;
    let n = years * 12;
    principal * r / (1.0 - (1.0 + r).powf(-(n as f64)))
}
```

A helper shows money with two decimals and thousands separators.

```rune
fn money(x) {
    let cents = (x * 100.0).round() as i64;
    let whole = `${cents / 100}`;
    let out = "";
    for i in 0..whole.len() {
        if i > 0 && (whole.len() - i) % 3 == 0 { out += ","; }
        out += whole[i..i + 1];
    }
    format!("{}.{:02}", out, cents % 100)
}
```

The loan: 300,000 at 6% a year for 30 years. The cell's value is the unrounded payment.

```rune
let (principal, rate, years) = (300000.0, 0.06, 30);
let pay = payment(principal, rate, years);
pay
```

```rune
println!("Loan {} at {:.1}% a year for {} years, paid monthly", money(principal), rate * 100.0, years);
println!("(a constant rate; totals exclude fees, taxes and insurance)");
println!("Monthly payment   {:>12}", money(pay));
println!("(totals use the unrounded payment, shown here to four decimals: {:.4})", pay);
println!("Total paid        {:>12}", money(pay * 360.0));
println!("Total interest    {:>12}", money(pay * 360.0 - principal));
println!("");
```

Each month, interest accrues on the balance and the rest of the payment pays it down.

```rune
println!("Balance after each year:");
let balance = principal;
for month in 1..=years * 12 {
    balance -= pay - balance * rate / 12.0;
    let year = month / 12;
    if month % 12 == 0 && [1, 5, 10, 20, 30].iter().any(|y| y == year) {
        let shown = if balance.abs() < 0.005 { 0.0 } else { balance };
        println!("  year {:>2}  {:>12}", year, money(shown));
    }
}
```
